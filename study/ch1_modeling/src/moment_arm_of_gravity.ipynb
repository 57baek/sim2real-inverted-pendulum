```python
import os
from datetime import datetime
import numpy as np  # type: ignore
import matplotlib  # type: ignore
matplotlib.use("Agg")
import matplotlib.pyplot as plt  # type: ignore
from matplotlib.patches import Arc, FancyArrowPatch  # type: ignore

RESULT_DIR = os.path.join(os.getcwd(), "result")
os.makedirs(RESULT_DIR, exist_ok=True)
TIMESTAMP = datetime.now().strftime("%Y%m%d_%H%M%S")

# ------------------------------------------------------------------
# STEP 1: set the rod and the tilts we want to draw
# What we have : rod length L (m), mass m (kg), gravity g (m/s^2)
# What we want : the moment arm d = (L/2) sin(theta) and the torque U0 sin(theta)
# Why          : to SEE that gravity (vertical) turns the rod through the
#                HORIZONTAL gap between the pivot and the line of the weight
# Formulas     : U0 = m g L / 2,   d = (L/2) sin(theta),   torque = m g d = U0 sin(theta)
# ------------------------------------------------------------------
rod_length = 1.0
rod_mass = 1.0
gravity = 9.81
energy_scale = rod_mass * gravity * rod_length / 2.0  # U0 in N*m
tilt_degrees_list = [0.0, 30.0, 60.0]

# ------------------------------------------------------------------
# STEP 2: draw one panel per tilt
# What we have : tilt theta from upright
# What we want : picture of rod, weight, line of the weight, moment arm
# Why          : same construction at three tilts shows the arm growing
# ------------------------------------------------------------------
fig, axes = plt.subplots(1, 3, figsize=(16, 6.4))
for axis, tilt_degrees in zip(axes, tilt_degrees_list):
    theta = np.radians(tilt_degrees)
    tip = np.array([rod_length * np.sin(theta), rod_length * np.cos(theta)])
    center = tip / 2.0
    arm = 0.5 * rod_length * np.sin(theta)
    torque = energy_scale * np.sin(theta)

    axis.plot([0, tip[0]], [0, tip[1]], color="#37474f", lw=9, solid_capstyle="round", zorder=2)
    axis.plot(0, 0, "o", color="black", ms=11, zorder=5)
    axis.plot(center[0], center[1], "o", color="#f9a825", ms=11, zorder=6)
    axis.plot([center[0], center[0]], [-0.5, 1.05], "--", color="#c62828", lw=1.3, zorder=1)
    axis.annotate("", xy=(center[0], center[1] - 0.42), xytext=(center[0], center[1]),
                  arrowprops=dict(arrowstyle="-|>", color="#c62828", lw=3), zorder=7)
    axis.text(center[0] + 0.04, -0.34, "weight m g, straight down\n(red dashed = its line)", color="#c62828", fontsize=10, va="top")
    if arm > 1e-9:
        axis.plot([0, 0], [0, 1.05], ":", color="gray", lw=1.3, zorder=1)
        axis.text(0.0, 1.09, "upright", color="gray", fontsize=10, ha="center")
        axis.annotate("", xy=(0, 0), xytext=(center[0], 0),
                      arrowprops=dict(arrowstyle="<->", color="#1565c0", lw=3.5), zorder=8)
        axis.text(center[0] / 2.0, -0.05, "moment arm d", color="#1565c0", fontsize=12,
                  ha="center", va="top", fontweight="bold")
        bracket_x = center[0] + 0.12
        axis.annotate("", xy=(bracket_x, 0), xytext=(bracket_x, center[1]),
                      arrowprops=dict(arrowstyle="<->", color="#2e7d32", lw=2), zorder=8)
        axis.text(bracket_x + 0.03, center[1] / 2.0, "height\n(energy,\nnot torque)", color="#2e7d32", fontsize=9, va="center")
        axis.add_patch(Arc((0, 0), 0.7, 0.7, theta1=90 - tilt_degrees, theta2=90, color="#6a1b9a", lw=2))
        axis.text(0.42 * np.sin(theta / 2.0), 0.42 * np.cos(theta / 2.0), "θ", color="#6a1b9a", fontsize=14, ha="center", va="center")
        axis.add_patch(FancyArrowPatch((-0.14, 0.07), (0.14, 0.07), connectionstyle="arc3,rad=-0.8",
                                       arrowstyle="-|>", mutation_scale=18, color="#ef6c00", lw=2.5))
        axis.text(-0.5, 0.2, "gravity turns the\nrod this way", color="#ef6c00", fontsize=10)
    else:
        axis.text(0.0, 1.09, "upright = line of the weight", color="gray", fontsize=10, ha="center")
        axis.text(0.06, -0.05, "moment arm d = 0", color="#1565c0", fontsize=12, va="top", fontweight="bold")
    axis.set_title(f"θ = {tilt_degrees:.0f}°:  d = (L/2) sin θ = {arm:.2f} m\n"
                   f"torque = m g d = {torque:.2f} N·m", fontsize=12)
    axis.set_xlim(-0.55, 1.0)
    axis.set_ylim(-0.6, 1.2)
    axis.set_aspect("equal")
    axis.axis("off")

fig.suptitle("Gravity is vertical, but it turns the rod through the HORIZONTAL gap (moment arm) between the pivot and the line of the weight",
             fontsize=13, y=0.985)
fig.tight_layout(rect=(0, 0, 1, 0.92))
figure_path = os.path.join(RESULT_DIR, f"moment_arm_gravity_{TIMESTAMP}.png")
fig.savefig(figure_path, dpi=150)
print("saved the figure")
plt.show()
```
